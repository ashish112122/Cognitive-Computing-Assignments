```python
#1
import numpy as np
arr = np.array([1, 2, 3, 4, 5])
print(arr + 2)
print(arr * 3)
print(arr / 2)
```

```python
#2
arr = np.array([1, 2, 3, 6, 4, 5]) 
arr = np.flip(arr)
print(arr)


x = np.array([1,2,3,4,5,1,2,1,1,1]) 
val, idx, count = np.unique(x, return_counts = True, return_inverse = True)

i = np.argmax(count)
print(val[i])

print(np.where(x == val[i]))

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ]) 

val, idx, count = np.unique(y, return_counts = True, return_inverse = True)

i = np.argmax(count)
print(val[i])

print(np.where(y == val[i]))
```

```python
#4

ucs420_ashish = np.array((10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35)).reshape(3, 4)

print(np.mean(ucs420_ashish))
print(np.max(ucs420_ashish))
print(np.min(ucs420_ashish))
print(np.unique(ucs420_ashish))

reshaped_ucs420_ashish = ucs420_ashish.reshape(4, 3)
print(reshaped_ucs420_ashish, '\n')

resized_ucs420_ashish = np.resize(ucs420_ashish, (2,3))
print(resized_ucs420_ashish)
```
