```python
#Q-3
    
original = np.array([1, 2, 3, 4, 5, 6])

subset = original[1:4]
subset[0] = 999

print("Original:", original)
print("Subset:", subset)

copied = original[1:4].copy()
copied[0] = 500

print("Original:", original)
print("Copied:", copied)

matrix = np.arange(1, 13).reshape(3, 4)

print("Matrix:")
print(matrix)

print("First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3:")
print(matrix[0:2, 1:3])

flat = matrix.flatten()
ravelled = matrix.ravel()

print("Flatten:", flat)
print("Ravel:", ravelled)

ravelled[0] = 999
print("After modifying ravel:")
print(matrix)

matrix = np.arange(1, 13).reshape(3, 4)

flat = matrix.flatten()
flat[0] = 500

print("After modifying flatten:")
print(matrix)

print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Dtype:", matrix.dtype)
```

```python
# Q4
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("Shape:", X.shape)
print("Dimensions:", X.ndim)

print("Transpose:")
print(X.T)

XtX = X.T @ X
print("X.T @ X:")
print(XtX)

inverse = np.linalg.inv(XtX)
print("Inverse:")
print(inverse)

beta = inverse @ X.T @ y
print("Coefficients:", beta)

new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta

print("Predicted assistance score:", predicted_score)
```
