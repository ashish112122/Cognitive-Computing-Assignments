```python
#2

def score_query(query, df):
    query_words = set(query.lower().split())
    results = []
    for i, row in df.iterrows():
        keywords = set(row["keywords"].lower().split())
        question_words = set(row["question"].lower().split())
        score = len(query_words & keywords) + len(query_words & question_words)

        if score > 0:
            results.append((score, i, row["question"]))

    results.sort(reverse=True)

    print("\nMatching Entries: ")
    for score, i, question in results:
        print(f"Score: {score} | {question}")


query = input("\nEnter your query ")
score_query(query, df)
```

```python

def same_category(category_name, df):
    return df[df["category"] == category_name]["question"]

print("Questions in account category")
print(same_category("account", df))

```

```python

def score_query_with_ties(query, df):
    query_words = set(query.lower().split())
    results = []

    for i, row in df.iterrows():
        keywords = set(row["keywords"].lower().split())
        question_words = set(row["question"].lower().split())

        score = len(query_words & keywords) + len(query_words & question_words)

        if score > 0:
            results.append((score, row["question"]))

    if not results:
        print("No matching entries.")
        return

    max_score = max(score for score, question in results)

    print(f"\nQuery: {query}")
    print(f"Highest score: {max_score}")

    for score, question in results:
        if score == max_score:
            print(f"Score =  {score},  {question}")


score_query_with_ties("fee", df)

```
