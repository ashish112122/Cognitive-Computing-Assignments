```python
df = pd.read_csv(r"C:\Users\HP\Desktop\Cognitive\Asg 3\employees.csv")
print("Shape\n", df.shape, '\n')
print("Info\n", df.info(), '\n')
print("Desc\n", df.describe(), '\n')
print(df.head(5), df.tail(5))
print(df["Salary"].mean())
print(df["Bonus"].sum())
print(df["Age"].min())
print(df["Rating"].max())
df = df.sort_values(by = "Salary", ascending = False)
print(df)
def performance(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"
df["Performance"] = df["Rating"].apply(performance)
print(df)
print(df.isnull().sum())
df = df.rename(columns = {"Employee_ID" : "ID"})
print(df[df["Years_of_Experience"] > 5])
print(df[df["Department"] == "IT"])
df["Tax"] = df["Salary"] * 0.10
print(df)
```

```python
df.to_csv("employees_modified.csv", index=False)
```
