```python
# Q -1 

roll  = 1024170341

l = list()

while roll:
    l.append((roll%10)*10)
    roll = roll // 10

l = l[::-1]
print(l)

l.append(50)
print(l, "added 50 at the end using append")
l.insert(0, 50)
print(l, "added 50 at the beginning of the list using insert")


l.pop()
print(l)

l.remove(50)
print(l)

l.sort()
print(l)

l.sort(reverse= True)
print(l)

print(l[:3])
print(l[-3:])

avg = sum(l)/ len(l)
newList = [i for i in l if i > avg]
print(newList)
```

```python
#Q-3
import random as r
r.seed(1024170341)

ran  = [r.randint(100,900) for i in range(100)]
print(ran)

count = 0
for i in ran:
    if i % 2 == 0:
        print(i, end = " ")
        count += 1
print("\n Total Even = ", count)

count = 0
for i in ran:
    if i % 2 != 0:
        print(i, end = " ")
        count += 1
print("\n Total Odd = ", count)


def prime(n):
    for i in range(2, int(n**(1/2))):
        if n % i == 0:
            return False
    return True

count = 0
for i in ran:
    if prime(i):
        print(i, end = " ")
        count += 1
print("\n Total Prime = ", count)

prime_num = [i for i in ran if prime(i)]
print(prime_num)


res = ran[0]
d = dict()
d[ran[0]] = 1
maxCount = 1
res = ran[0]
for i in ran[1:]:
    d[i] = d.get(i,0) + 1
    if maxCount < d[i]:
        maxCount = d[i]
        res = i
print("Most Freq = ", res)
print("Its count =", count)





```

```python
#Q - 4

rollno = [1,0,2,4,1,7,0,3,4,1]
A = {i*7 for i in rollno[:8]}
print(A)

B = {i*9 for i in rollno[:8]}
print(B)

print(A.union(B))

print(A.intersection(B))

print(A.difference(B))
print(B.difference(A))

print(A.symmetric_difference(B), " difference is elements in A whihc do not exist in B whule symmetric difference is A U B - A U B")

print(A.issubset(B))
print(B.issuperset(A))

x = int(input("Enter the numbe to remove"))
A.discard(x)
print(A)
print("discard is safer because it does not raise an error if the element is not in the set unline remove()")

```
